```python
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
import matplotlib.pyplot as plt
```

```python
X = [
    [30, 70],
    [25, 90],
    [28, 80],
    [35, 60],
    [20, 95]
]
```

```python
y = [1, 0, 1, 1, 0]
```

```python
model = DecisionTreeClassifier()
```

```python
model.fit(X, y)
```

```python
prediction = model.predict([[29, 75]])

print("Prediction:", prediction)
```

```python
tree.plot_tree(model, filled=True)
plt.show()
```

```python
# Input (X) and Output (Y)
X = [1, 2, 3, 4, 5]
Y = [20000, 25000, 30000, 35000, 40000]
```

```python
# Mean calculate
mean_x = sum(X) / len(X)
mean_y = sum(Y) / len(Y)

print(mean_x, mean_y)
```

```python
# Calculate slope (m)
num = 0
den = 0

for i in range(len(X)):
    num += (X[i] - mean_x) * (Y[i] - mean_y)
    den += (X[i] - mean_x) ** 2

m = num / den
print("Slope:", m)
```

```python
# Calculate intercept (c)
c = mean_y - (m * mean_x)
print("Intercept:", c)
```

```python
# Prediction
x_new = 6
y_pred = m * x_new + c

print("Predicted Salary:", y_pred)
```
